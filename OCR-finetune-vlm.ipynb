````python
import os
from os.path import join
import random
from tqdm.auto import tqdm
from glob import glob
from pdf2image import convert_from_path
from PIL import Image
from PIL import ImageEnhance
data_dir = "/home/abood/coding/MAWRID/data"
````

````python
def preprocess_images(image, maxwidth=1200):
    # keep RGB — vision models expect 3-channel input
    rgb = image.convert('RGB')

    if rgb.width > maxwidth:
        aspect_ratio = maxwidth / rgb.width
        new_height = int(rgb.height * aspect_ratio)
        rgb = rgb.resize((maxwidth, new_height), Image.LANCZOS)

    enhancer = ImageEnhance.Contrast(rgb)
    enhanced = enhancer.enhance(1.5)

    return enhanced


def convert_pdf_to_images(pdf_path, output_dir, maxwidth=1200):
    pdf_name = os.path.basename(pdf_path).replace('.pdf', '')
    output_dir = join(output_dir, pdf_name)
    os.makedirs(output_dir, exist_ok=True)

    print(f"Processing {pdf_name}.pdf ...")

    images = convert_from_path(pdf_path, dpi=200)
    generated_paths = []

    for page_num, image in enumerate(images, start=1):
        preprocessed_image = preprocess_images(image, maxwidth)
        output_path = join(output_dir, f"{pdf_name}_page_{page_num:03d}.jpeg")
        preprocessed_image.save(output_path, 'JPEG', quality=90, optimize=True)
        generated_paths.append(output_path)
        print(f"Saved page {page_num} -> {output_path}")

    return generated_paths
````

````python
output_dir = f"{data_dir}/processed_images" 
pdf_files = glob(f"{data_dir}/raw_pdfs/*.pdf")
maxwidth = 1200

for pdf_file in tqdm(pdf_files, desc="Processing PDFs"):
    convert_pdf_to_images(pdf_file, output_dir, maxwidth)
````

# MAWRID — 3-Stage Ollama Pipeline Tester
Tests Stage 1 (OCR), Stage 2 (classification), Stage 3 (field extraction) separately with local Ollama models.

| Stage | Model | Task |
|-------|-------|------|
| 1 — OCR | `gemma3:4b` (vision) | image → raw_text |
| 2 — Classify | `qwen2.5:3b` or `gemma3:4b` | raw_text → doc_type |
| 3 — Extract | `qwen2.5:3b` or `gemma3:4b` | raw_text + doc_type → fields |

````python
import json, re, base64, time, textwrap
from pathlib import Path
from openai import OpenAI

REPO_ROOT   = Path('/home/abood/coding/MAWRID')
SCHEMA_PATH = REPO_ROOT / 'backend/config/schema_v2.json'
IMAGES_DIR  = REPO_ROOT / 'data/processed_images'
OLLAMA_BASE = 'http://localhost:11434'

# Models to compare (edit freely)
VISION_MODELS = ['qwen2.5vl:3b']              # document OCR — dynamic tiling
TEXT_MODELS   = ['qwen2.5:3b', 'gemma3:4b']   # both run on GPU ✓

# Document to test
DOC_ID = '0001'

SCHEMA = json.loads(SCHEMA_PATH.read_text(encoding='utf-8'))
FALLBACK_TYPE = 'غير_محدد'
print('Schema loaded —', len(SCHEMA['documents']), 'document types')
print('Available docs:', sorted(p.name for p in IMAGES_DIR.iterdir()))
````

## Prompt builders — identical to MAWRID backend

````python
def _type_list_compact():
    lines = []
    for key, doc in SCHEMA['documents'].items():
        label    = doc.get('label_ar', key)
        category = doc.get('category_label_ar') or doc.get('category', '')
        lines.append(f'• {key} — {label} ({category})')
    return '\n'.join(lines)

def build_stage1_prompt():
    return '''أنت نظام OCR متخصص في استخراج النصوص من الوثائق العربية الرسمية.

## المهمة
استخرج كل النص المرئي في الصورة بدقة كاملة:
- الحفاظ على النص بخطه الأصلي (عربي أو إنجليزي) بدون ترجمة أو تحويل
- الحفاظ على فواصل الأسطر والفقرات
- تضمين النصوص في الأختام والطوابع والتوقيعات والهوامش
- الأرقام والتواريخ كما هي بالضبط

## صيغة الإخراج
أعد JSON فقط — لا نص قبله ولا بعده — يبدأ بـ { وينتهي بـ }:
{
  "raw_text": "النص الكامل المستخرج من الوثيقة"
}'''

def build_stage1_prompt_plain():
    return '''أنت نظام OCR. استخرج كل النص المرئي في الصورة كما هو — عربي أو إنجليزي — مع الحفاظ على تنسيق الأسطر والفقرات والأرقام بدقة تامة. لا تضف تفسيرات أو تعليقات.'''


def build_stage2_prompt():
    type_list = _type_list_compact()
    return f'''أنت نظام تصنيف وثائق دقيق.

## المهمة
اقرأ النص أدناه وحدد نوع الوثيقة من القائمة التالية فقط.
أعد المفتاح بالضبط كما هو مكتوب (مثال: تقرير_طبي وليس "تقرير طبي").

## أنواع الوثائق
{type_list}

إذا لم يتطابق النص مع أي نوع اذكر: {FALLBACK_TYPE}

## صيغة الإخراج
أعد JSON فقط — لا نص قبله ولا بعده:
{{
  "document_type": "المفتاح_بالضبط",
  "confidence": "high | medium | low"
}}

## النص:
'''

def build_stage3_prompt(doc_type):
    doc_def = SCHEMA['documents'].get(doc_type)
    if not doc_def:
        raise ValueError(f'Unknown doc type: {doc_type}')
    label  = doc_def['label_ar']
    fields = doc_def['fields']
    field_lines = []
    for f in fields:
        fid, flabel, ftype = f['id'], f['label_ar'], f['type']
        req = ' [مطلوب]' if f.get('required') else ' [اختياري]'
        if ftype == 'lookup':
            opts = ' | '.join(f.get('options', []))
            hint = f'اختر قيمة واحدة بالضبط من: {opts}'
        elif ftype == 'date':
            hint = 'تاريخ بصيغة YYYY-MM-DD فقط'
        elif ftype == 'number':
            hint = 'رقم صحيح فقط'
        else:
            hint = 'نص'
        field_lines.append(f'  "{fid}": null  // {flabel}{req} — {hint}')
    fields_block = '\n'.join(field_lines)
    return f'''أنت نظام استخراج بيانات دقيق من الوثائق الرسمية.

## نوع الوثيقة: {label}

## قواعد الاستخراج
- أعد كل النصوص بخطها الأصلي بدون أي ترجمة
- إذا لم يوجد الحقل في النص أعد null
- lookup: أعد إحدى القيم المذكورة حرفياً
- date: أعد التاريخ بصيغة YYYY-MM-DD دائماً
- number: أعد رقماً صحيحاً فقط

## الحقول المطلوبة
أعد JSON فقط — لا نص قبله ولا بعده:
{{
{fields_block}
}}

## النص:
'''

print('Prompt builders ready')
````

## Ollama call helpers

````python
import ollama

NUM_CTX = 4096  # cap context to fit KV cache in 8GB VRAM (default 128k causes CPU fallback)

def call_vision_multipage(doc_id, model, prompt):
    pages = sorted((IMAGES_DIR / doc_id).glob('*.jpeg'))
    t0 = time.monotonic()
    resp = ollama.chat(
        model=model,
        messages=[{
            'role': 'user',
            'content': prompt,
            'images': [str(p) for p in pages],
        }],
        options={'num_ctx': NUM_CTX},
    )
    ms = round((time.monotonic() - t0) * 1000, 1)
    return {
        'text': resp.message.content or '',
        'latency_ms': ms,
        'input_tokens': resp.prompt_eval_count or 0,
        'output_tokens': resp.eval_count or 0,
    }

def call_vision_single(image_path, model, prompt):
    t0 = time.monotonic()
    resp = ollama.chat(
        model=model,
        messages=[{
            'role': 'user',
            'content': prompt,
            'images': [str(image_path)],
        }],
        options={'num_ctx': NUM_CTX},
    )
    ms = round((time.monotonic() - t0) * 1000, 1)
    return {
        'text': resp.message.content or '',
        'latency_ms': ms,
        'input_tokens': resp.prompt_eval_count or 0,
        'output_tokens': resp.eval_count or 0,
    }

def call_text(prompt, model):
    t0 = time.monotonic()
    resp = ollama.chat(
        model=model,
        messages=[{'role': 'user', 'content': prompt}],
        options={'num_ctx': NUM_CTX},
    )
    ms = round((time.monotonic() - t0) * 1000, 1)
    return {
        'text': resp.message.content or '',
        'latency_ms': ms,
        'input_tokens': resp.prompt_eval_count or 0,
        'output_tokens': resp.eval_count or 0,
    }

def parse_json(text):
    cleaned = re.sub(r'```(?:json)?\s*', '', text).strip().rstrip('`').strip()
    cleaned = re.sub(r'//[^\n"]*', '', cleaned)
    try:
        return json.loads(cleaned)
    except json.JSONDecodeError:
        m = re.search(r'\{.*\}', cleaned, re.DOTALL)
        if m:
            try: return json.loads(m.group())
            except: pass
    return {}

print('Helpers ready — num_ctx:', NUM_CTX)
````

## Stage 1 — OCR (vision model)
Send document image(s) → get raw Arabic text back.

````python
stage1_results = {}
s1_prompt = build_stage1_prompt()

for model in VISION_MODELS:
    print(f'\n>>> Stage 1 — {model} — doc {DOC_ID} ...')
    try:
        r = call_vision_multipage(DOC_ID, model, s1_prompt)
        parsed = parse_json(r['text'])
        raw_text = parsed.get('raw_text', r['text'])
        raw_text = raw_text.replace('\\n', '\n')  # small models double-escape newlines in JSON
        stage1_results[model] = {**r, 'raw_text': raw_text}
        print(f'    latency: {r["latency_ms"]:.0f} ms  |  tokens in/out: {r["input_tokens"]}/{r["output_tokens"]}')
        print('─'*60)
        print(raw_text[:800] + ('...' if len(raw_text) > 800 else ''))
    except Exception as e:
        print(f'    ERROR: {e}')
        stage1_results[model] = {'raw_text': '', 'latency_ms': 0, 'error': str(e)}
````

## Stage 1b — Quick single-page test
Useful for checking quality on just one page without sending all pages.

````python
PAGE_TO_TEST = 1  # change to test a different page

RAW_IMAGES_DIR = REPO_ROOT / 'data/processed_images'

pages = sorted((RAW_IMAGES_DIR / DOC_ID).glob('*.jpeg'))
print(f'Doc {DOC_ID} has {len(pages)} page(s):', [p.name for p in pages])

raw_1b = ''
stage1b_model = VISION_MODELS[0]

if pages:
    target = pages[PAGE_TO_TEST - 1]
    for model in VISION_MODELS:
        print(f'\n>>> Stage 1b — {model} — {target.name}')
        r = call_vision_single(target, model, build_stage1_prompt_plain())
        raw = r['text'].strip()
        raw_1b = raw
        stage1b_model = model
        print(f'    {r["latency_ms"]:.0f} ms  |  tokens in: {r["input_tokens"]}  out: {r["output_tokens"]}')
        print('─' * 60)
        print(raw[:800] + ('...' if len(raw) > 800 else ''))
````

## Stage 2b — Classify single-page result
Uses the `raw_text` captured in Stage 1b above.

````python
if not raw_1b.strip():
    print('No raw_text — run Stage 1b first')
else:
    print(f'Using raw_text from Stage 1b ({len(raw_1b)} chars)\n')
    s2_prompt = build_stage2_prompt()
    stage2b_results = {}

    for model in TEXT_MODELS:
        print(f'>>> Stage 2b — {model} ...')
        try:
            r = call_text(s2_prompt + raw_1b, model)
            parsed   = parse_json(r['text'])
            doc_type = parsed.get('document_type', FALLBACK_TYPE)
            conf     = parsed.get('confidence', 'low')
            stage2b_results[model] = {**r, 'doc_type': doc_type, 'confidence': conf}
            print(f'    → {doc_type}  (conf: {conf})  |  {r["latency_ms"]:.0f} ms')
        except Exception as e:
            print(f'    ERROR: {e}')
            stage2b_results[model] = {'doc_type': FALLBACK_TYPE, 'confidence': 'low', 'error': str(e)}

    print('\n── Classification summary ────────────────────────')
    for m, res in stage2b_results.items():
        print(f'  {m:30s}  {res["doc_type"]}  ({res["confidence"]})  {res.get("latency_ms",0):.0f}ms')
````

## Stage 3b — Extract fields from single-page result
Uses `raw_1b` + `doc_type` from Stage 2b.

````python
if 'stage2b_results' not in dir() or not stage2b_results:
    print('Run Stage 2b first')
else:
    S2B_SOURCE  = TEXT_MODELS[0]
    doc_type_1b = stage2b_results.get(S2B_SOURCE, {}).get('doc_type', FALLBACK_TYPE)

    # Override manually if needed:
    # doc_type_1b = 'شهادة_ميلاد'

    print(f'Using doc_type: {doc_type_1b}  (from {S2B_SOURCE})')

    if doc_type_1b not in SCHEMA['documents']:
        print(f'WARNING: "{doc_type_1b}" not in schema — cannot run Stage 3b')
        print('Tip: set doc_type_1b manually above and re-run')
    else:
        fields_def = SCHEMA['documents'][doc_type_1b]['fields']
        print(f'Schema defines {len(fields_def)} fields\n')
        s3_prompt  = build_stage3_prompt(doc_type_1b)
        stage3b_results = {}

        for model in TEXT_MODELS:
            print(f'>>> Stage 3b — {model} ...')
            try:
                r = call_text(s3_prompt + raw_1b, model)
                fields = parse_json(r['text'])
                stage3b_results[model] = {**r, 'fields': fields}
                print(f'    latency: {r["latency_ms"]:.0f} ms')
                print(json.dumps(fields, ensure_ascii=False, indent=2))
                print()
            except Exception as e:
                print(f'    ERROR: {e}')
                stage3b_results[model] = {'fields': {}, 'error': str(e)}

        # Side-by-side comparison
        if len(stage3b_results) > 1:
            all_keys = set()
            for res in stage3b_results.values():
                all_keys.update(res.get('fields', {}).keys())
            models = list(stage3b_results.keys())
            w = 35
            print('── Field comparison ──────────────────────────────')
            header = f'{"Field":<30}' + ''.join(f'{m[:w-1]:<{w}}' for m in models)
            print(header)
            print('─' * len(header))
            for key in sorted(all_keys):
                row = f'{key:<30}'
                for m in models:
                    val = stage3b_results[m].get('fields', {}).get(key, '—')
                    row += f'{str(val)[:w-2]:<{w}}'
                print(row)
````

## Stage 2 — Classification
Send raw_text from Stage 1 → identify document type from 91 options.

````python
S1_SOURCE = VISION_MODELS[0]
raw_text = stage1_results.get(S1_SOURCE, {}).get('raw_text', '')

if not raw_text.strip():
    print('No raw_text — run Stage 1 cell first')
else:
    s2_prompt = build_stage2_prompt()
    stage2_results = {}

    for model in TEXT_MODELS:
        print(f'\n>>> Stage 2 — {model} ...')
        try:
            r = call_text(s2_prompt + raw_text, model)
            parsed   = parse_json(r['text'])
            doc_type = parsed.get('document_type', FALLBACK_TYPE)
            conf     = parsed.get('confidence', 'low')
            stage2_results[model] = {**r, 'doc_type': doc_type, 'confidence': conf}
            print(f'    → {doc_type}  (conf: {conf})  |  {r["latency_ms"]:.0f} ms')
        except Exception as e:
            print(f'    ERROR: {e}')
            stage2_results[model] = {'doc_type': FALLBACK_TYPE, 'confidence': 'low', 'error': str(e)}

    print('\n── Classification summary ─────────────────────────')
    for m, res in stage2_results.items():
        print(f'  {m:30s}  {res["doc_type"]}  ({res["confidence"]})  {res.get("latency_ms",0):.0f}ms')
````

## Stage 3 — Field Extraction
Send raw_text + doc_type → extract structured fields defined in schema.

````python
S2_SOURCE = TEXT_MODELS[0]
doc_type  = stage2_results.get(S2_SOURCE, {}).get('doc_type', FALLBACK_TYPE)

# Override manually if needed:
# doc_type = 'هوية_شخصية'

print(f'Using doc_type: {doc_type}')

if doc_type not in SCHEMA['documents']:
    print(f'WARNING: "{doc_type}" not in schema — cannot run Stage 3')
else:
    fields_def = SCHEMA['documents'][doc_type]['fields']
    print(f'Schema defines {len(fields_def)} fields for this type\n')

    s3_prompt = build_stage3_prompt(doc_type)
    stage3_results = {}

    for model in TEXT_MODELS:
        print(f'>>> Stage 3 — {model} ...')
        try:
            r = call_text(s3_prompt + raw_text, model)
            fields = parse_json(r['text'])
            stage3_results[model] = {**r, 'fields': fields}
            print(f'    latency: {r["latency_ms"]:.0f} ms')
            print(json.dumps(fields, ensure_ascii=False, indent=2))
            print()
        except Exception as e:
            print(f'    ERROR: {e}')
            stage3_results[model] = {'fields': {}, 'error': str(e)}
````

## Field comparison — models side by side

````python
if 'stage3_results' not in dir() or not stage3_results:
    print('Run Stage 3 first')
else:
    all_keys = set()
    for res in stage3_results.values():
        all_keys.update(res.get('fields', {}).keys())

    models = list(stage3_results.keys())
    w = 35
    header = f'{"Field":<30}' + ''.join(f'{m[:w-1]:<{w}}' for m in models)
    print(header)
    print('─' * len(header))
    for key in sorted(all_keys):
        row = f'{key:<30}'
        for m in models:
            val = stage3_results[m].get('fields', {}).get(key, '—')
            row += f'{str(val)[:w-2]:<{w}}'
        print(row)
````

## Batch — run full pipeline on all 15 docs
Saves results to `mawrid_bench/results/`.

````python
BATCH_VISION = VISION_MODELS[0]
BATCH_TEXT   = TEXT_MODELS[0]
out_file = REPO_ROOT / f'mawrid_bench/results/results_ollama_{BATCH_VISION.replace(":","_")}.json'
out_file.parent.mkdir(parents=True, exist_ok=True)

s1p = build_stage1_prompt()
s2p = build_stage2_prompt()
all_docs = sorted(p.name for p in IMAGES_DIR.iterdir() if p.is_dir())
print(f'Batch: {len(all_docs)} docs  vision={BATCH_VISION}  text={BATCH_TEXT}')

batch = {}
for doc_id in all_docs:
    print(f'[{doc_id}] ', end='', flush=True)
    e = {'doc_id': doc_id, 'vision_model': BATCH_VISION, 'text_model': BATCH_TEXT}

    try:
        r1 = call_vision_multipage(doc_id, BATCH_VISION, s1p)
        p1 = parse_json(r1['text'])
        raw = p1.get('raw_text', r1['text'])
        e['raw_text']  = raw.replace('\\n', '\n')  # small models double-escape newlines
        e['stage1_ms'] = r1['latency_ms']
        print(f'S1✓{r1["latency_ms"]:.0f}ms ', end='', flush=True)
    except Exception as ex:
        e.update({'raw_text':'','stage1_ms':0,'stage1_error':str(ex)})
        print('S1✗'); batch[doc_id]=e; continue

    try:
        r2 = call_text(s2p + e['raw_text'], BATCH_TEXT)
        p2 = parse_json(r2['text'])
        e['doc_type']   = p2.get('document_type', FALLBACK_TYPE)
        e['confidence'] = p2.get('confidence', 'low')
        e['stage2_ms']  = r2['latency_ms']
        print(f'S2✓{r2["latency_ms"]:.0f}ms ', end='', flush=True)
    except Exception as ex:
        e.update({'doc_type':FALLBACK_TYPE,'stage2_ms':0,'stage2_error':str(ex)})
        print('S2✗'); batch[doc_id]=e; continue

    if e['doc_type'] not in SCHEMA['documents']:
        e.update({'fields':{},'stage3_ms':0})
        print('S3-skip')
    else:
        try:
            r3 = call_text(build_stage3_prompt(e['doc_type']) + e['raw_text'], BATCH_TEXT)
            e['fields']    = parse_json(r3['text'])
            e['stage3_ms'] = r3['latency_ms']
            print(f'S3✓{r3["latency_ms"]:.0f}ms')
        except Exception as ex:
            e.update({'fields':{},'stage3_ms':0,'stage3_error':str(ex)})
            print('S3✗')

    e['total_ms'] = e.get('stage1_ms',0)+e.get('stage2_ms',0)+e.get('stage3_ms',0)
    batch[doc_id] = e

out_file.write_text(json.dumps(batch, ensure_ascii=False, indent=2), encoding='utf-8')
print(f'\nSaved → {out_file}')
````

## Latency summary

````python
if 'batch' not in dir() or not batch:
    print('Run batch cell first')
else:
    print(f'{"Doc":<8}  {"Type":<32}  {"S1(ms)":>8}  {"S2(ms)":>8}  {"S3(ms)":>8}  {"Total":>8}')
    print('─' * 80)
    totals = []
    for doc_id, e in batch.items():
        s1,s2,s3,tot = e.get('stage1_ms',0),e.get('stage2_ms',0),e.get('stage3_ms',0),e.get('total_ms',0)
        totals.append(tot)
        print(f'{doc_id:<8}  {e.get("doc_type","?"):<32}  {s1:>8.0f}  {s2:>8.0f}  {s3:>8.0f}  {tot:>8.0f}')
    avg = sum(totals)/len(totals) if totals else 0
    print(f'\nAverage total: {avg:.0f} ms  ({avg/1000:.1f}s)')
````
